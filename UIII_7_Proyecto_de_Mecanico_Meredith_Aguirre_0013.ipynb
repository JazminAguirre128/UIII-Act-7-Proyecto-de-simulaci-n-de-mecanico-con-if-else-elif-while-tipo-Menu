```python
print("==========================================")
print("       MECÁNICO INTELIGENTE AUTOBOT       ")
print("==========================================")
continuar = "si"
while continuar == "si":
    print("\n--- MENÚ DE DIAGNÓSTICO ---")
    print("1. Revisar fallas del motor")
    print("2. Revisar sistema de frenos")
    print("3. Salir del simulador")
    opcion = input("Selecciona la opción deseada (1, 2 o 3): ")
    if opcion == "1":
        temperatura = float(input("¿A cuántos grados (Celsius) está el motor?: "))
        print("La temperatura registrada es de:", temperatura, "grados.")
        if temperatura > 100:
            print(" IA Dice: ¡Alerta! El motor está sobrecalentado. Apágalo de inmediato.")
        else:
            print(" IA Dice: Temperatura normal. El sistema de enfriamiento funciona bien.")
    elif opcion == "2":
        vida_frenos = float(input("¿Qué porcentaje (%) de vida útil les queda a las balatas?: "))
        print("A tus frenos les queda el:", vida_frenos, "% de vida.")
        if vida_frenos < 20:
            print(" IA Dice: ¡Peligro! Tus frenos están muy desgastados. Necesitas cambiarlos ya.")
        else:
            print(" IA Dice: Frenos seguros. Tienen suficiente vida para seguir viajando.")
    elif opcion == "3":
        print(" Sistema apagado. ¡Conduce con cuidado!")
        continuar = "no"
    else:
        print(" Opción incorrecta. Elige 1, 2 o 3.")
    if opcion != "3":
        continuar = input("\n¿Quieres consultar otra falla? (si/no): ")
print("==========================================")
print("Programa realizado por Meredith Aguirre NL = 2")
```
